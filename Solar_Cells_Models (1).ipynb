# **All** **Libraries**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import r2_score, accuracy_score, mean_absolute_error
from sklearn.preprocessing import LabelEncoder
import tensorflow as tf
from tensorflow.keras import layers, Model
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from tensorflow.keras.callbacks import EarlyStopping
import shap
import matplotlib.pyplot as plt
```

# **R F model**


```python
physics_features = ['L', 'mob_IL', 'mob_IR', 'mun_0', 'mup_0', 'IL_dop', 'IR_dop', 'Etrap', 'Cn', 'Cp', 'num_GBs', 'GB_tr']
X = df[physics_features]
y_efficiency = df['FF1.00']
y_faults = LabelEncoder().fit_transform(df['label'])
```

```python
X_train_full, X_test_full, y_train_eff_full, y_test_eff_full, y_train_fault_full, y_test_fault_full = train_test_split(
    X, y_efficiency, y_faults, test_size=0.2, random_state=42
)
```

```python
n_splits = 5
X_splits = np.array_split(X_train_full, n_splits)
y_eff_splits = np.array_split(y_train_eff_full, n_splits)
y_fault_splits = np.array_split(y_train_fault_full, n_splits)

regressors = []
classifiers = []
```

```python
for i in range(n_splits):
    print(f"{i+1} {n_splits}...")
```

```python
    rf_reg = RandomForestRegressor(n_estimators=50, max_depth=8, min_samples_leaf=10, n_jobs=-1, random_state=42)
    rf_reg.fit(X_splits[i], y_eff_splits[i])
    regressors.append(rf_reg)
```

```python
    rf_clf = RandomForestClassifier(n_estimators=50, n_jobs=-1, random_state=42)
    rf_clf.fit(X_splits[i], y_fault_splits[i])
    classifiers.append(rf_clf)
```

```python
eff_preds = np.array([reg.predict(X_test_full) for reg in regressors])
fault_preds = np.array([clf.predict(X_test_full) for clf in classifiers])
```

```python
final_eff_pred = np.mean(eff_preds, axis=0)
final_fault_pred = np.round(np.mean(fault_preds, axis=0)).astype(int)
```

```python
r2 = r2_score(y_test_eff_full, final_eff_pred)
mae = mean_absolute_error(y_test_eff_full, final_eff_pred)
acc = accuracy_score(y_test_fault_full, final_fault_pred)
```

```python
print(f"R² (Effectivity): {r2*100:.2f}%")
print(f"MAE (Effectivity): {mae:.4f}")
print(f"Accuracy (Faults): {acc*100:.2f}%")
```

```python
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(16, 6))


sns.scatterplot(x=y_test_eff_full, y=final_eff_pred, ax=axes[0], alpha=0.5, color='darkblue', s=40)
axes[0].plot([y_test_eff_full.min(), y_test_eff_full.max()], [y_test_eff_full.min(), y_test_eff_full.max()], 'r--', lw=2.5, label='Ideal Fit')
axes[0].set_title(' Ensemble Model: True vs Predicted (FF1.00)', fontsize=13, fontweight='bold')
axes[0].set_xlabel('True Values (Actual Physics)', fontsize=11)
axes[0].set_ylabel('Predicted Values (Model Forecast)', fontsize=11)
axes[0].legend()

mean_importances = np.mean([reg.feature_importances_ for reg in regressors], axis=0)
indices = np.argsort(mean_importances)[::-1]

sns.barplot(x=mean_importances[indices], y=np.array(physics_features)[indices], ax=axes[1], palette='plasma')
axes[1].set_title(' Mean Feature Importance across Splits', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Mean Importance Score', fontsize=11)
axes[1].set_ylabel('Physics Features', fontsize=11)

plt.tight_layout()
plt.show()
```

#NN


```python
advanced_features = ['L', 'mob_IL', 'mob_IR', 'mun_0', 'mup_0', 'IL_dop', 'IR_dop', 'Etrap', 'Cn', 'Cp', 'num_GBs', 'GB_tr', 'Gmax', 'Rseries', 'Rshunt', 'n']
X = df[advanced_features]
y_efficiency = df['FF1.00']

le = LabelEncoder()
y_faults = le.fit_transform(df['label'])
num_classes = len(np.unique(y_faults))
```

```python
X_train_full, X_test_full, y_train_eff, y_test_eff, y_train_fault, y_test_fault = train_test_split(
    X, y_efficiency, y_faults, test_size=0.2, random_state=42
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_full)
X_test_scaled = scaler.transform(X_test_full)
```

```python
inputs = layers.Input(shape=(X_train_scaled.shape[1],), name='advanced_input')

shared = layers.Dense(256, activation='relu')(inputs)
shared = layers.BatchNormalization()(shared)
shared = layers.Dropout(0.2)(shared)

shared = layers.Dense(128, activation='relu')(shared)
shared = layers.BatchNormalization()(shared)
shared = layers.Dropout(0.2)(shared)

shared = layers.Dense(64, activation='relu')(shared)
shared = layers.BatchNormalization()(shared)
```

```python
eff_head = layers.Dense(32, activation='relu')(shared)
eff_output = layers.Dense(1, activation='linear', name='efficiency_output')(eff_head)

fault_head = layers.Dense(32, activation='relu')(shared)
fault_output = layers.Dense(num_classes, activation='softmax', name='fault_output')(fault_head)

model = Model(inputs=inputs, outputs=[eff_output, fault_output])
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss={'efficiency_output': 'mse', 'fault_output': 'sparse_categorical_crossentropy'},
    metrics={'efficiency_output': 'mae', 'fault_output': 'accuracy'}
)
```

```python
early_stop = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
history = model.fit(
    X_train_scaled,
    {'efficiency_output': y_train_eff, 'fault_output': y_train_fault},
    epochs=50,
    batch_size=256,
    validation_split=0.1,
    callbacks=[early_stop],
    verbose=1
)
```

```python
eval_results = model.evaluate(X_test_scaled, {'efficiency_output': y_test_eff, 'fault_output': y_test_fault}, verbose=0)
preds = model.predict(X_test_scaled, verbose=0)
r2_dnn = r2_score(y_test_eff, preds[0].flatten())

print(f"(R2 Score): {r2_dnn * 100:.2f}%")
print(f" (Accuracy): {eval_results[4] * 100:.2f}%")
```

```python
def predict_eff_only(x_data):
    preds = model.predict(x_data, verbose=0)
    return preds[0].flatten()

background_samples = X_train_scaled[:100]
test_samples_to_explain = X_test_scaled[:50]

explainer = shap.Explainer(predict_eff_only, background_samples)

shap_values = explainer(test_samples_to_explain)

plt.figure(figsize=(12, 8))
shap.summary_plot(
    shap_values.values,
    test_samples_to_explain,
    feature_names=advanced_features,
    show=False
)

plt.title(' SHAP Global Explanation: Feature Impact on Efficiency (FF1.00)', fontsize=14, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()
```
